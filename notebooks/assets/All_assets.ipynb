```python
import yfinance as yf
import numpy as np

tickers = ['SPY', 'QQQ', 'IWM', 'EEM', 'EWJ', 'EWQ', 'TLT', 'GLD']

returns = {}
for t in tickers:
    ret = yf.download(t, start='2000-01-01', end='2024-12-31',
                      auto_adjust=True, progress=False)['Close'].squeeze().pct_change().dropna()
    returns[t] = ret

for ticker in tickers:
    corrs = []
    for other in tickers:
        if other == ticker:
            continue
        common = returns[ticker].index.intersection(returns[other].index)
        corrs.append(returns[ticker].loc[common].corr(returns[other].loc[common]))
    print(f"{ticker} — mean corr panel: {np.mean(corrs):.3f}")
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
import numpy as np

X = df[['vol', 'mean_corr_panel']].values
y_xgb = df['mean_f1_xgb'].values
y_rfc = df['mean_f1_rfc'].values

from scipy.stats import pearsonr

df['score_composite'] = df['vol'] * df['mean_corr_panel'].clip(lower=0)

pr_xgb, pv_xgb = pearsonr(df['score_composite'], df['mean_f1_xgb'])
pr_rfc, pv_rfc = pearsonr(df['score_composite'], df['mean_f1_rfc'])

print(f"Composite vs XGB — r={pr_xgb:.3f} p={pv_xgb:.3f}")
print(f"Composite vs RFC — r={pr_rfc:.3f} p={pv_rfc:.3f}")

from numpy.linalg import lstsq
X_scaled = (X - X.mean(axis=0)) / X.std(axis=0)
X_aug = np.column_stack([X_scaled, np.ones(len(X_scaled))])

coef_xgb, _, _, _ = lstsq(X_aug, y_xgb, rcond=None)
coef_rfc, _, _, _ = lstsq(X_aug, y_rfc, rcond=None)

y_pred_xgb = X_aug @ coef_xgb
y_pred_rfc = X_aug @ coef_rfc

ss_res_xgb = ((y_xgb - y_pred_xgb)**2).sum()
ss_tot_xgb = ((y_xgb - y_xgb.mean())**2).sum()
ss_res_rfc = ((y_rfc - y_pred_rfc)**2).sum()
ss_tot_rfc = ((y_rfc - y_rfc.mean())**2).sum()

print(f"\n Multivariate regression (vol + corr_panel)")
print(f"R² XGB : {1 - ss_res_xgb/ss_tot_xgb:.3f}")
print(f"R² RFC : {1 - ss_res_rfc/ss_tot_rfc:.3f}")
print(f"\nCoefficients XGB (vol, corr_panel, intercept) : {coef_xgb.round(3)}")
print(f"Coefficients RFC (vol, corr_panel, intercept) : {coef_rfc.round(3)}")
```
